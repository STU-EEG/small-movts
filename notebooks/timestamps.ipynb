```python
import matplotlib.pyplot as plt
from datetime import datetime

def parse_timestamp(timestamp_str):
    """
    Parse timestamp strings in two formats:
    1. 'DD-MMM-YYYY_HHhMMmSS.sss' format
    2. 'DD HH:MM:SS' format (where day can be 27, 28, etc.)
    """
    if not timestamp_str or timestamp_str.strip() == '':
        return None
    
    try:
        # Format 1: '27-MAY-2025_23h44m33.873s'
        if '_' in timestamp_str and 'h' in timestamp_str:
            # Remove trailing 's' if present
            clean = timestamp_str.rstrip('s')
            # Replace the underscore and time markers
            clean = clean.replace('_', ' ')
            clean = clean.replace('h', ':')
            clean = clean.replace('m', ':')
            # Parse: '27-MAY-2025 23:44:33.873'
            return datetime.strptime(clean, '%d-%b-%Y %H:%M:%S.%f')
        
        # Format 2: '27 23:49:53'
        elif ' ' in timestamp_str and ':' in timestamp_str:
            # This format doesn't include month/year, so we need to infer from context
            # For now, we'll handle it differently - return None and handle separately
            return None
        
        else:
            return None
            
    except (ValueError, AttributeError):
        return None

def parse_relative_timestamp(timestamp_str, reference_date):
    """
    Parse 'DD HH:MM:SS' format relative to a reference date
    """
    if not timestamp_str or timestamp_str.strip() == '':
        return None
    
    parts = timestamp_str.split()
    if len(parts) == 2:
        day = int(parts[0])
        time_parts = parts[1].split(':')
        hour = int(time_parts[0])
        minute = int(time_parts[1])
        second = int(time_parts[2]) if len(time_parts) > 2 else 0
        
        # Create datetime using reference year/month and the specified day
        # Assume same month as reference, but day might roll over
        ref_year = reference_date.year
        ref_month = reference_date.month
        
        # Handle day rollover (if day < reference day, it's next month)
        if day < reference_date.day:
            # Next month
            if ref_month == 12:
                ref_month = 1
                ref_year += 1
            else:
                ref_month += 1
        
        return datetime(ref_year, ref_month, day, hour, minute, second)
    
    return None

def timestamp_to_ms(dt, start_time):
    """Convert datetime to milliseconds offset from start_time"""
    if dt is None:
        return None
    delta = dt - start_time
    return delta.total_seconds() * 1000

def process_timestamps(timestamps_map):
    """Process the timestamps map and extract key/value pairs in milliseconds"""
    keys_ms = []
    values_ms = []
    
    # First, get all valid keys to determine start time
    valid_keys = []
    for key, value in timestamps_map.items():
        dt_key = parse_timestamp(key)
        if dt_key:
            valid_keys.append((key, dt_key, value))
    
    if not valid_keys:
        print("No valid keys found")
        return [], []
    
    # Use the earliest key as reference start time
    start_time = min(dt for _, dt, _ in valid_keys)
    
    for key, dt_key, value in valid_keys:
        # Convert key to ms
        key_ms = timestamp_to_ms(dt_key, start_time)
        
        # Parse value (could be empty string or 'DD HH:MM:SS' format)
        if value and value.strip():
            dt_value = parse_relative_timestamp(value, dt_key)
            if dt_value:
                value_ms = timestamp_to_ms(dt_value, start_time)
            else:
                # Try parsing as absolute timestamp
                dt_value_abs = parse_timestamp(value)
                if dt_value_abs:
                    value_ms = timestamp_to_ms(dt_value_abs, start_time)
                else:
                    print(f"Could not parse value: '{value}' for key: '{key}'")
                    continue
        else:
            # print(f"Skipping empty value for key: '{key}'")
            continue
        
        keys_ms.append(key_ms)
        values_ms.append(value_ms)
    
    return keys_ms, values_ms


# Process the data
keys_ms, values_ms = process_timestamps(timestamps_map)



plt.figure(figsize=(12, 8))

# Plot the points
plt.plot(keys_ms, values_ms, 'bo-', markersize=6, linewidth=2, label='Data points')

# Add identity line (y=x) for reference
max_val = max(max(keys_ms), max(values_ms))
min_val = min(min(keys_ms), min(values_ms))
plt.plot([min_val, max_val], [min_val, max_val], 'r--', alpha=0.5, label='y=x (perfect alignment)')

# Labels and title
plt.xlabel('EEG time (milliseconds from start)', fontsize=12)
plt.ylabel('Video Timestamp (milliseconds from start)', fontsize=12)
# plt.title('Timestamps: Values vs Keys (offset from start)', fontsize=14)

plt.grid(True, alpha=0.3)
plt.legend()
plt.gca().ticklabel_format(style='plain', axis='both')
plt.tight_layout()
plt.show()
```
