```python
fa = face_alignment.FaceAlignment(face_alignment.LandmarksType.TWO_D, flip_input=False, device="cpu")
```

```python
cap = cv2.VideoCapture(str(video_path))
fps = cap.get(cv2.CAP_PROP_FPS) or 30
w = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
h = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

output_cropped_path = data_dir / "head_cropped_face_aligned.mp4"
fourcc = cv2.VideoWriter_fourcc(*"avc1")
writer = cv2.VideoWriter(str(output_cropped_path), fourcc, fps, (w, h), isColor=True)


# 1m 15s
while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    vis_frame = frame.copy()

    preds = fa.get_landmarks_from_image(frame, return_bboxes=True)
    if preds and preds[0]:
        for detection in preds[0]:
            for point in detection:
                cv2.circle(vis_frame, (int(point[0]), int(point[1])), 2, (255, 0, 0), -1)  # Blue circles for landmarks

        x1, y1, x2, y2 = preds[2][0][:4]
        cv2.rectangle(vis_frame, (int(x1), int(y1)), (int(x2), int(y2)), (0, 255, 0), 2)  # Green rectangle
    writer.write(vis_frame)

cap.release()
writer.release()
```
